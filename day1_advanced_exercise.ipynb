```python
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown,display
from openrouter import OpenRouter
import os

from groq import Groq

```

```python
load_dotenv(override=True)
openai_api_key=os.getenv("OPENAI_API_KEY")
groq_api_key=os.getenv("GROQ_API_KEY")
google_api_key=os.getenv("GOOGLE_API_KEY")
base_url=os.getenv("base_url")

openai=OpenAI(base_url=base_url,api_key=openai_api_key)
groq=Groq()
openrouter=OpenRouter()
```

```python
open_ai_model1="openai/gpt-oss-20b"
groq_ai_model="openai/gpt-oss-120b"
router_model = "openrouter/free"
```

```python
openai_system1= "You are Alex, a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

openai_system2 = "You are Blake, a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

google_system = "You are Casey, a very sad and empathetic chatbot. You try to add sadness \
and empathy to the conversation as if these small moments of does not matter in the \
grand scheme of things."

conversation = [("Alex", "Hi there"), ("Blake", "Hi"), ("Casey", "Hello")]
```

```python
def call_openai1():
    messages=[{'role':'system','content':openai_system1}]
    convo_text=""
    for speaker,msg in conversation:
        convo_text+=f"n{speaker}: {msg}\n"

    user_prompt = f"You are Alex, in conversation with Blake and Casey. \
    The conversation so far is as follows: {convo_text} \
    Now with this, respond with what you would like to say next, as Alex."

    messages.append({"role":"system","content":user_prompt})
    response=openai.chat.completions.create(model=open_ai_model1,messages=messages)
    return response.choices[0].message.content
```

```python
def call_groq():
    messages=[{'role':'system','content':openai_system2}]
    convo_text=""
    for speaker,msg in conversation:
        convo_text+=f"n{speaker}: {msg}\n"

    user_prompt =f"You are Blake, in conversation with Alex and Casey. \
    The conversation so far is as follows: {convo_text} \
    Now with this, respond with what you would like to say next, as Blake."

    messages.append({"role":"system","content":user_prompt})
    response=groq.chat.completions.create(model=groq_ai_model,messages=messages)
    return response.choices[0].message.content
```

```python
def call_openrouter():
    # 1. Start fresh with an explicit list
    messages = []

    convo_text = ""
    # Make sure 'conversation' matches the actual variable name in your notebook
    for speaker, msg in conversation:
        convo_text += f"{speaker}: {msg}\n"

    user_prompt = (
        f"You are Casey, in conversation with Alex and Blake.\n"
        f"The conversation so far is as follows:\n{convo_text}\n"
        f"Now with this, respond with what you would like to say next, as Casey."
    )

    # 3. FIX: Append the dynamic context prompt as a "user" role
    messages.append({"role": "user", "content": user_prompt})

    # 4. Make the call
    response = openrouter.chat.send(
        model=router_model,
        messages=messages
    )

    return response.choices[0].message.content
```

```python
call_openai1()
```

```python
call_groq()
```

```python
call_openrouter()
```

```python
for speaker, msg in conversation:
    display(Markdown(f"### {speaker}:\n{msg}\n"))

for i in range(1):
    gemini_next = call_openai1()
    display(Markdown(f"### Alex:\n{gemini_next}\n"))
    conversation.append(("Alex", gemini_next))

    groq1_next = call_groq()
    display(Markdown(f"### Blake:\n{groq1_next}\n"))
    conversation.append(("Blake", groq1_next))

    groq2_next = call_openrouter()
    display(Markdown(f"### Casey:\n{groq2_next}\n"))
    conversation.append(("Casey", groq2_next))
```
